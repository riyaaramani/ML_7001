```python
import pandas as pd
import numpy as ns
```

```python
df=pd.read_csv("02 computer.csv")
df
```

```python
df['age']=df['age'].str.replace('31...40','31-40')
df
```

```python
x=df.iloc[:,0:4]
y=df['buys_computer']
x
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=1
)
```

```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

oe_age = OrdinalEncoder(categories=[['<=30', '31-40', '>40']])
oe_income = OrdinalEncoder(categories=[['low', 'medium', 'high']])
oe_credit_rating = OrdinalEncoder(categories=[['fair', 'excellent']])

x_train['age'] = oe_age.fit_transform(x_train[['age']])
x_train['income'] = oe_income.fit_transform(x_train[['income']])
x_train['credit_rating'] = oe_credit_rating.fit_transform(x_train[['credit_rating']])

x_test['age'] = oe_age.transform(x_test[['age']])
x_test['income'] = oe_income.transform(x_test[['income']])
x_test['credit_rating'] = oe_credit_rating.transform(x_test[['credit_rating']])

ohe = OneHotEncoder(drop='if_binary', sparse_output=False)

x_train[['student_yes']] = ohe.fit_transform(x_train[['student']])
x_test[['student_yes']] = ohe.transform(x_test[['student']])

x_train = x_train.drop('student', axis=1)
x_test = x_test.drop('student', axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()

y_train=le.fit_transform(y_train)
y_test=le.transform(y_test)
```

```python
x_train
```

```python
from sklearn.naive_bayes import CategoricalNB
```

```python
model = CategoricalNB()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score

ac = accuracy_score(y_test, y_pred)
print(f"Accuracy: {ac:.3%}")
```
